```python
#check hot encoding for gender and marital status
train_df_cat = train_df_all.select_dtypes(exclude=["number"])
train_df_cat.fillna("Unknown", inplace = True)
# List of Columns before Encoding
initial_columns = train_df_cat.columns
from sklearn import preprocessing
le = preprocessing.LabelEncoder()

for i in range(len(train_df_cat.columns)):
    print(train_df_cat.columns[i])
    if train_df_cat.columns[i]=="CODE_GENDER":
        train_df_cat[str(train_df_cat.columns[i]) + "_LE"] = le.fit_transform(train_df_cat[train_df_cat.columns[i]])
        print(le.inverse_transform([0,1]))
    elif train_df_cat.columns[i]=="NAME_FAMILY_STATUS":
        train_df_cat[str(train_df_cat.columns[i]) + "_LE"] = le.fit_transform(train_df_cat[train_df_cat.columns[i]])
        # print(train_df_cat["NAME_FAMILY_STATUS_LE"].unique())
        print(le.inverse_transform([0,1,2,3,4]))
    
```
